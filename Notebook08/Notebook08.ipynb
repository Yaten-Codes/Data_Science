# Exercise 4: Identify Popular Movies

Next we would like to only consider movies that have a significant number of ratings. This task is complicated by the fact that movies and ratings are in 2 different DataFrames and we want to filter the `movies` DataFrame based on a statistic from the `ratings` DataFrame.

First let's compute the number of ratings per movie:

```python
number_of_ratings = ratings.movieId.value_counts()
number_of_ratings.head()
```

Now we want to filter this pandas Series object and keep only the rows where the count is larger or equal to 100.
We don't want to pollute our analysis with movies with a tiny number of reviews. Call the filtered series "number_of_ratings_of_popular_movies".

```python
number_of_ratings_of_popular_movies = None
### BEGIN SOLUTION
### YOUR CODE HERE!
number_of_ratings_of_popular_movies = number_of_ratings[number_of_ratings >= 100]

### END SOLUTION
```

```python
assert len(number_of_ratings_of_popular_movies) == 10326, "Try again, check that movies with 100 ratings are accepted"
```

Finally we want to use the `reindex` function to change the index of movies, this will create a new DataFrame with a new index that contains the movieId of only the most popular movies.

The value of all the rows of `movies` that have the same movieId will be copied over to the new `all_popular_movies` dataframe, the rest will be discarded.

```python
all_popular_movies = movies.reindex(number_of_ratings_of_popular_movies.index).sort_index()
all_popular_movies.head()
```

# Exercise 5: Filter by Genre

Implement a general function that filters movies by genres. Include the movie as long as the genre is at least one of the generes mentioned in the genres column of the dataframe. It does not have to be an exact match to the entire genres entry for that movie. HINT: Your solution should use str.contains. The function should return the filtered dataframe.

```python
def filter_by_genre(input_movies, genre):
    """Return only movies of a specific genre"""
    ### BEGIN SOLUTION
    ### YOUR CODE HERE!
    return input_movies[input_movies.genres.str.contains(genre)]


    ### END SOLUTION
```

Then let's apply it to the `popular_movies` dataset to retain only the "Fantasy" movies:

```python
fantasy_movies = filter_by_genre(all_popular_movies, "Fantasy")
```

```python
assert len(fantasy_movies) == 742, """Try again, Make sure you are filtering the popular movies"""
```

# Exercise 6: Join Movies and Ratings

Let's create a single `DataFrame` that contains both titles and mean ratings of the popular fantasy movies.

Titles are only available in the `movies` `DataFrame`, while ratings are only available in the `ratings` `DataFrame`. We would like to create a single DataFrame that includes the title and the mean rating.

First create the `mean_ratings` dataframe by computing the mean rating for each movieId:

```python
mean_ratings = ratings[['movieId','rating']].groupby('movieId').mean()
mean_ratings.head()
```

```python
fantasy_movies.head()
```

Now properly merge the mean_ratings and fantasy_movies dataframes to create a new "fantasy_movies_with_ratings" dataframe. There should be one extra column of 'rating' added to the fantasy_movies dataframe after the merge.

```python
fantasy_movies_with_ratings = None
### BEGIN SOLUTION
### YOUR CODE HERE!
fantasy_movies_with_ratings = pd.merge(fantasy_movies, mean_ratings, on='movieId')
### END SOLUTION
```
